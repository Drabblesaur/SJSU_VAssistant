```python
Registrar_Links = [
    'https://www.sjsu.edu/registrar/academic-records/change-of-major-minor.php',
    'https://www.sjsu.edu/registrar/academic-records/grade-changes.php',
    'https://www.sjsu.edu/registrar/academic-records/transcripts.php',
    'https://www.sjsu.edu/registrar/academic-records/update-personal-information.php',
    'https://www.sjsu.edu/registrar/registration/grading-option.php',
    'https://www.sjsu.edu/registrar/academic-records/inadmissible-repeat.php',
    'https://www.sjsu.edu/registrar/registration/registration-basics/index.php',
    'https://www.sjsu.edu/registrar/registration/repeat-grade-forgiveness.php',
]
```

```python
def scrape_links(links):
    data = []
    for link in links:
        print(f"Processing {link}")
        response = requests.get(link)
        soup = BeautifulSoup(response.text, 'html.parser')
        main_content = soup.find('main', id='sjsu-maincontent')
        if not main_content:
            print(f"Main content not found for {link}")
            continue
        # Extract text from main_content
        content_text = main_content.get_text()
        # Save link and content as a dict into list
        data.append({'link': link, 'content': content_text})
    return data
```

```python
registrar_data = scrape_links(Registrar_Links)
```
